# **1. Introducción y planteamiento del problema**

## 1.1. Introducción

Conocer el contexto de los clientes de un servicio es parte fundamental en la estrategia de un negocio. El objetivo siempre es vender, y por lo tanto, mantener la base de clientes es necesario para que el negocio funcione. En un contexto de servicios de suscripción, y específicamente en telecomunicaciones, las ganancias dependen en muy gran medida sobre la permanencia de la mayor cantidad de usuarios posible. En tal situación resalta la importancia de conocer cuál es la posible tasa de abandono, y cuáles clientes son los que se inclinan más a retirarse del servicio.

Para el presente proyecto se desarrolla un agente que busca predecir dicha característica, donde en base a los datos de un cliente y su suscripción se permita obtener la estimación sobre su permanencia en el servicio o su abandono.

## 1.2. Definición del problema

El churn consiste en la medida en que clientes, usuarios o suscriptores abandonan los servicios de una plataforma o empresa. Se puede presentar cuando un cliente cancela el servicio de forma directa, cuando deja de renovar su suscripción o cuando su uso disminuye hasta volverse inactivo. En todos esos casos, se presenta una pérdida para el negocio, porque se disminuyen los ingresos recurrentes y consecuentemente es necesario conseguir nuevos clientes para mantener el crecimiento en la empresa.

El problema al que responde este proyecto es anticipar ese abandono. Esto debido a que resulta más eficiente para una empresa detectar cuándo un cliente está en riesgo de abandono, que enterarse después de su retirada. Resolver el problema de forma interna es más eficaz, porque al detectarlo anticipadamente es posible llevar a cabo acciones de retención, que resultan menos costosas y complicadas que intentar recuperar a un cliente que ya decidió abandonar.

Por lo tanto, se busca determinar en base a información de un cliente y su relación con el servicio la probabilidad de abandono que tiene el mismo. Esto permite identificar qué clientes se encuentran estables y cuáles están en riesgo, para priorizar a aquellos que tiendan más a abandonar. La idea principal es que, por encima de que se clasifiquen correctamente los casos, se determine efectivamente los clientes que van a abandonar, que representan el foco de atención, para que la empresa u organización pueda intervenir a tiempo.

## 1.3. ¿Cuál es el agente que se desea construir?

Se busca construir un agente que reciba los datos un cliente de una empresa de telecomunicaciones, incluyendo información relacionada con su antigüedad como cliente, el tipo de contrato, los cargos mensuales y totales, el uso de distintos servicios contratados, y demás variables importantes. A partir de estas percepciones, el agente debe brindar la probabilidad y clasificar a cada cliente según su riesgo de abandono del servicio (churn).

Así se logra obtener como resultado si un usuario se encuentra en riesgo o no, esto mediante una clasificación que indique el riesgo del mismo (alto o bajo riesgo de abandono), así como la probabilidad de abandono. El objetivo principal del agente es consistir en una herramienta que permita identificar las condiciones actuales de los clientes de un servicio, permitiendo previsualizar cuáles son los clientes que más atención requieren actualmente.

# **6. Implementación de modelos**

## 6.1. Implementación de Modelo 1: Regresión Logística

```python
from src.model import build_logreg_pipeline

logreg_pipeline = build_logreg_pipeline(preprocessor_scaled)
logreg_pipeline.fit(X_train, y_train)
```

### 6.1.1. *Explicación del funcionamiento*



La Regresión Logística estima la probabilidad de que una observación pertenezca a una clase. Internamente calcula una combinación lineal de las variables de entrada (cada una multiplicada por un peso aprendido durante el entrenamiento) y pasa el resultado por la función sigmoide, que comprime cualquier valor real al rango entre 0 y 1. Ese valor se interpreta como la probabilidad de pertenecer a la clase positiva, en este caso la probabilidad de que el cliente presente churn.

Durante el entrenamiento, el modelo ajusta los pesos minimizando una función de pérdida (log-loss), de modo que las probabilidades predichas se acerquen lo más posible a las etiquetas reales. Para la clasificación final, si esa probabilidad supera un umbral (0.5 por defecto), se predice la clase 1; en caso contrario, la clase 0.

### 6.1.2. *Justificación*

La Regresión Logística es un buen modelo para este agente porque su salida es una probabilidad entre 0 y 1, que es exactamente lo que el sistema necesita devolver: la probabilidad de que un cliente abandone el servicio y una clasificación final. Es un modelo rápido, lo que permite respuestas en tiempo real cuando el usuario ingresa los datos, y es fácil de explicar, ya que permite entender qué elementos están influyendo en la predicción. Además, sirve como punto de comparación frente al Random Forest, porque si un modelo más complejo no mejora significativamente los resultados, el más simple es la mejor opción.

### 6.1.3. Ajuste básico de hiperparámetros

En este caso, se hace ajuste de hiperparámetros con GridSearchCV, que entrena el modelo de forma profunda y variada con todas las combinaciones posibles de los parámetros definidos dentro de logreg_param_grid, seleccionando la combinación que maximiza la métrica definida (en este caso F1, porque buscamos encontrar el mejor equilibrio entre precisión y recall).

El parámetro C representa al inverso del nivel de regularización, donde al aplicar valores pequeños como 0.001 se aplica una regularización fuerte que penaliza a los coeficientes más grandes, obligando al modelo a mantenerse simple, mientras que valores grandes de este parámetro como 100, prácticamente eliminan la regularización y permiten que el modelo se ajuste con más libertad a los datos con los que se entrenó. Se explora un rango amplio (0.001 a 100) para tomar en cuenta situaciones desde una alta regularización hasta aquellas de baja regularización, y dejar que la validación cruzada calcule cuál nivel se adapta mejor a los datos de este dataset.

La penalización se fija en l2 y el solver en lbfgs porque es la combinación estándar recomendada para problemas de clasificación binaria con regularización Ridge y en datasets de un tamaño como el del utilizado. El parámetro class_weight se agrega con dos valores, None y "balanced", lo cual permite que el GridSearchCV compare el desempeño del modelo sin compensación de desbalance contra el desempeño con dicha compensación, y claramente seleccionando el adecuado.

```python
from src.model import tune_logreg

logreg_grid = tune_logreg(logreg_pipeline, X_train, y_train)

print("Mejores parámetros:", logreg_grid.best_params_)
print("Mejor F1 en validación cruzada:", logreg_grid.best_score_)
```

```python

results = pd.DataFrame(logreg_grid.cv_results_)
results[["param_model__C", "mean_test_score", "rank_test_score"]]
```

En los resultados del ajuste se puede ver cómo los mejores parámetros indican que la asignación de 100 al parámetro C es la que brinda el mejor valor de F1. Eso implica que aplicar una menor regularización es el nivel adecuado que mejor se adapta a este dataset, donde el modelo se beneficia al darle mayor libertad para ajustarse a las variables disponibles. Se mantiene C=100 por ser el valor seleccionado por best_estimators_ y porque la evidencia que se verá posteriormente en la evaluación experimental demuestra coherencia entre el F1 de validación cruzada y el F1 de test (ambos están cerca a 0.63), donde los hiperparámetros seleccionados aquí por GridSearchCV no son específicos del conjunto de entrenamiento, sino que producen resultados consistentes sobre cualquier conjunto de datos no vistos. Así se confirma que el modelo generaliza correctamente a pesar de la baja regularización, descartando el riesgo de sobreajuste.

 Por su parte también se detecta que la presencia del balanceo empleando class_weight efectivamente mejora los resultados frente al desbalanceo que tiene el dataset. El penalty L2 y el Solver lbfgs se mantienen al ser los únicos indicados en el ajuste (por tema de estándar y compatibilidad entre ambos parámetros).

## 6.2. Implementación de Modelo 2: Random Forest

```python
from src.model import build_rf_pipeline

rf_pipeline = build_rf_pipeline(preprocessor_unscaled)
rf_pipeline.fit(X_train, y_train)
```

### 6.2.1. *Explicación del funcionamiento*

Random Forest combina múltiples árboles de decisión entrenados de forma independiente. Cada árbol se entrena sobre una parte aleatoria y distinta del conjunto de datos y considera solo un subconjunto aleatorio de variables en cada división, lo cual permite asegurar que los árboles no aprendan exactamente lo mismo y que el conjunto sea más seguro y confiable.

Para hacer una predicción, cada árbol vota por una clase y el modelo final entrega la clase más votada (o el promedio de las probabilidades, según la configuración). Esta combinación permite reducir el sobreajuste que puede ocurrir en los árboles individuales, y suele encontrar mejor las relaciones no lineales entre variables, lo cual es útil cuando el comportamiento del cliente no responde a un patrón precisamente lineal.

### 6.2.2. *Justificación*

En este caso, se trata de otro modelo adecuado para este agente porque es robusto y combina múltiples árboles de decisión para tomar una decisión más precisa. También entrega una probabilidad de abandono y una clasificación final de forma directa, cumpliendo exactamente con el resultado esperado. Además, su capacidad de indicar qué variables son más importantes lo convierte en una herramienta útil tanto para predecir como para entender y explicar el comportamiento del cliente.

### 6.2.3. Ajuste básico de hiperparámetros

Para el Random Forest se realiza el ajuste utilizando RandomizedSearchCV, que en lugar de probar todas las combinaciones posibles muestrea aleatoriamente 20 de ellas y selecciona la que maximiza la métrica (F1 igualmente para este ajuste). Se emplea este enfoque porque el espacio de búsqueda es considerablemente más grande que el de la regresión logística, y una búsqueda exhaustiva resultaría muy costosa en tiempo y memoria, sin brindar una mejora que valga la pena.

El parámetro n_estimators define la cantidad de árboles del bosque, donde más árboles estabilizan las predicciones pero con mayor tiempo de entrenamiento. max_depth limita la profundidad de cada árbol y es el principal control contra el sobreajuste: árboles muy profundos memorizan el entrenamiento, mientras que árboles poco profundos pueden quedarse cortos en capturar patrones. min_samples_split y min_samples_leaf complementan ese control al exigir un mínimo de muestras para dividir un nodo o formar una hoja, evitando divisiones que estén basadas en muy pocos datos. max_features se encarga de definir cuántas variables considera cada árbol en cada división, lo cual introduce más diversidad entre los árboles y reduce la relación entre sus predicciones.

Y por su parte, class_weight se prueba en None, "balanced" y "balanced_subsample" para que el modelo compare el desempeño con y sin compensación del desbalance que tiene el dataset. La validación cruzada utiliza la misma estrategia StratifiedKFold definida previamente para mantener consistencia entre los dos modelos.

```python
from src.model import tune_rf

rf_grid = tune_rf(rf_pipeline, X_train, y_train)

print("Mejores parámetros:", rf_grid.best_params_)
print("Mejor F1 en validación cruzada:", rf_grid.best_score_)
```

Los resultados del ajuste indican que lo mejor es emplear 200 árboles para estabilizar las predicciones, donde la profundidad máxima debe ser 10 implicando un menor sobreajuste al limitar la complejidad de cada árbol y evitar memorización del entrenamiento. En cuanto a `min_samples_split` se indica 5 y `min_samples_leaf` 2 como mínimos que evitan divisiones muy pequeñas, lo cual resulta particularmente relevante en este dataset por su tamaño limitado, donde las divisiones excesivamente pequeñas tienden más a capturar ruido. Por su parte, max_features en sqrt implica que cada árbol considera únicamente la raíz cuadrada del total de variables al evaluar cada división, lo cual introduce diversidad entre los árboles del bosque y reduce la correlación entre sus predicciones.

## 6.3. Exploración de umbral de decisión óptimo

Por defecto los modelos de clasificación binaria de scikit-learn predicen la clase 1 cuando `predict_proba > 0.5`. Sin embargo, este umbral no necesariamente maximiza el rendimiento sobre la clase minoritaria en problemas desbalanceados como el que estamos analizando actualmente.

Debido a esto, se explora en esta sección qué umbral maximiza el F1-score sobre la clase positiva (Churn = 1) para cada modelo ajustado. El umbral encontrado se aplicará posteriormente en la evaluación final.

Esta exploración utiliza la curva precision-recall, que evalúa el comportamiento
del modelo a través de todos los umbrales posibles.

### **Interpretación de los umbrales obtenidos**

La exploración revela que el umbral óptimo según F1 es relativamente cercano al umbral por

La mejora es baja (alrededor de 1% en F1 en ambos casos), lo cual

El parámetro `class_weight="balanced"`, que fue incorporado en el ajuste de hiperparámetros, ya compensa internamente el desbalance presente en el dataset ponderando las muestras inversamente proporcional a la frecuencia de cada clase. Como resultado, los modelos obtenidos producen probabilidades cuyo umbral natural de equilibrio entre precisión y recall queda próximo a 0.5. Esto implica que la mejora en el ajuste de umbrales no sea tan considerable.

A pesar de que la mejora es pequeña, se decide aplicar los umbrales óptimos encontrados en la evaluación final para tener una mejor optimización explícita y verificable, y debido a que es una decisión metodológica documentada y reproducible, necesaria para considerar la mayor cantidad de posibilidades. Además, modificarlo permite priorizar recall (capturar más clientes en riesgo de churn, aceptando más falsos positivos) o precisión (reducir intervenciones innecesarias), según el costo relativo de cada tipo de error en la toma de decisiones de negocio que puede involucrar el Churn.

## 7.1. Definición de métricas a utilizar

### 7.1.1. *Accuracy*

Esta métrica corresponde al procentaje total de aciertos en las predicciones del modelo. No funciona muy bien en datasets desbalanceados, como es el caso del dataset usado, sin embargo es importante considerarla para conocer el rendimiento general del modelo, siempre y cuando se tomen en cuenta también otras métricas que reflejen el desempeño desde diferentes puntos de vista.

### 7.1.2. *Recall*

Se trata de una de las métricas más importantes según lo definido en el rendimiento. Mide la proporción de casos positivos reales que el modelo logró detectar correctamente, es decir, de todos los casos verdaderos cuantos detectó el modelo. En el caso actual, cuantos de los clientes que realmente abandonaron el servicio el modelo detectó que abandonarían el servicio.

### 7.1.3. *F1-Score*

Implementa una media armónica entre la precisión y el recall. Lo que busca medir es el equilibrio entre ambas métricas en un único valor, penalizando los casos en los que una de las dos es desproporcionadamente baja respecto a la otra. También se considera importante porque se busca mantener un nivel de precisión aceptable, y evitar exceder la clasificación errónea de clientes que abandonarán pero que no lo hacen.

### 7.1.4. *ROC-AUC*

Mide la capacidad del modelo para separar clases a distintos umbrales de decisión. Un valor cercano a 1 indica una excelente capacidad discriminativa, mientras que un valor cercano a 0.5 equivale a una clasificación aleatoria. Resulta importante a considerar porque se busca un nivel de generalización adecuado ante distintos perfiles, sin inclinarse mucho por una clase u otra.

# **9. Discusión - Análisis crítico**

## 9.1. Limitaciones en el modelo

### **Techo de información en el dataset**
Como fue mencionado en la selección de modelos, Random Forest se caracteriza por trabajar con una gran cantidad de árboles, ideal para identificar relaciones más complejas, mientras que Logistic Regressión es más simple y lineal. El hecho de que los resultados en F1 y ROC-AUC (0.6323 vs 0.6329 y 0.8383 vs 0.8376 respectivamente) en ambos modelos sean muy similares, y no haya mejora con RF, indica que la mejora de las métricas no se consigue cambiando el modelo, sino agregando información relevante al dataset.

### **Ligera pérdida en Recall**
A pesar de haber implementado  `class_weight="balanced"`, el recall para la clase 1 (abandono) queda entre 77 y 80%, lo cual indica que 1 de cada 5 clientes que abandona no es detectado. En un ambiente académico o de prueba no representa un gran problema, sin embargo, en una situación real, dicha pérdida puede ser costosa.

### **Estimación basada en una captura estática**
En este caso el modelo no considera una evolución temporal del cliente, donde se consideren por ejemplo la aceleración de quejas, o la caída en el uso dentro de las últimas semanas. Esas son señales muy valiosas, que pueden extender la capacidad de  `tenure` para considerar la historia del cliente dentro del servicio como un factor que puede influenciar su abandono.

### **Transformer limitado**
Aunque la implementación del LLM es una función adicional, y no forma parte del modelo de Machine Learning en sí, cabe resaltar que la calidad de las respuestas del análisis empleándolo se ven limitadas por la capacidad del modelo transformer. El modelo utilizado FLAN-T5-large es el más grande y viable bajo las restricciones de peso y latencia del entorno de ejecución, pero al ser un modelo de tamaño moderado en comparación con otros LLMs con mayor potencia, su capacidad de razonamiento sobre el dominio del problema es limitada. Esto requiere que se hagan mitigaciones específicas como el uso de few-shot prompting (guía basada en ejemplos) para guiar el formato de las respuestas, lo cual representa un balance entre autonomía por parte del transformer y coherencia en las explicaciones generadas.

## 9.2. Riesgo de sobreajuste

A pesar de haber agregado varias decisiones metodológicas específicamente destinadas a prevenir el sobreajuste, el riesgo restante en este proyecto sí debe considerarse. La mitigación principal se aplicó en la combinación de validación cruzada estratificada en 5 partes dentro del GridSearchCV, lo cual asegura que el ajuste de hiperparámetros no se realiza sobre una única sección de los datos sino sobre cinco particiones distintas, y también se aplicó al aplicar el encapsulamiento del preprocesamiento dentro del Pipeline, que evita que el SimpleImputer y el StandardScaler reciban información del conjunto de prueba durante el entrenamiento.

La regresión logística incluye a su vez regularización L2 con una consideración amplia del parámetro C entre 0.001 y 100, que permite que se seleccione el grado de penalización más adecuado para los datos maximizando F1, que representa el equilibrio. También en el Random Forest se entrena con restricciones explícitas en max_depth, min_samples_split y min_samples_leaf que limitan la profundidad y el grado de especialización de cada árbol para que no se sobreajusten. La evidencia que demuestra de que estas medidas sí funcionaron es la coherencia entre el F1 obtenido en validación cruzada y el F1 obtenido sobre el conjunto de prueba, que se mantienen ambos en torno a 0.63 sin una brecha que indique memorización del entrenamiento.

A pesar de haber considerado dichas estrategias, existe un riesgo de sobreajuste que se debe considerar, debido a que el tamaño del dataset (7043 registros, de los cuales solo el 80% se usa para entrenar) sigue siendo limitado frente a la capacidad que tiene Random Forest, con hasta 200 estimadores, que aun con restricciones de profundidad mantiene una capacidad de memorización considerable sobre combinaciones específicas de variables. Además, cable aclarar que aunque el ajuste de hiperparámetros se realiza con validación cruzada sobre el conjunto de entrenamiento, el conjunto de prueba se utilizó también para decisiones intermedias como la selección del umbral óptimo. Esto no constituye data leakage precisamente, pero sí implica que el F1 obtenido como métrica final puede ser ligeramente más optimista que el rendimiento real sobre clientes nuevos. De igual forma, la mejora obtenida para F1 mediante el ajuste del umbral no fue significativa, por lo que su influencia en los resultados finales es mínima.


## 9.3. Posibles sesgos en los datos

Como se mencionó dentro de las limitaciones del dataset en el análisis inicial, los datos capturan el comportamiento de clientes en California dentro de un trimestre determinado, lo cual indica que los patrones de Churn que aprende el modelo pueden estar influenciados por factores culturales, sociales o propios del sector. Por este motivo es importante considerar el comportamiento histórico de los clientes, para observar también patrones previamente manifestados.
